```python
set_a = ["sqft_living", "bedrooms", "bathrooms"]
set_b = ["sqft_living", "sqft_above", "sqft_living15"]

C_a = np.corrcoef(df[set_a].to_numpy(float), rowvar=False)
C_b = np.corrcoef(df[set_b].to_numpy(float), rowvar=False)

det_a, det_b = np.linalg.det(C_a), np.linalg.det(C_b)
print("Correlation matrix A  [sqft_living, bedrooms, bathrooms]")
print(C_a, f"\n  det = {det_a:.4f}")
print("\nCorrelation matrix B  [sqft_living, sqft_above, sqft_living15]")
print(C_b, f"\n  det = {det_b:.4f}")
```

```python
eig_feats = ["sqft_living", "bedrooms", "bathrooms", "floors", "grade",
             "sqft_above", "sqft_living15", "yr_built"]

Xe = df[eig_feats].to_numpy(float)
Zs = (Xe - Xe.mean(axis=0)) / Xe.std(axis=0)        # standardized matrix
R = np.corrcoef(Xe, rowvar=False)                    # correlation matrix (8 x 8)

fig, ax = plt.subplots(figsize=(8.2, 6.4))
im = ax.imshow(R, cmap="viridis", vmin=-0.2, vmax=1)
ax.set_xticks(range(len(eig_feats))); ax.set_xticklabels(eig_feats, rotation=45, ha="right")
ax.set_yticks(range(len(eig_feats))); ax.set_yticklabels(eig_feats)
for i in range(len(eig_feats)):
    for j in range(len(eig_feats)):
        ax.text(j, i, f"{R[i, j]:.2f}", ha="center", va="center", fontsize=8.5,
                color="#f5f7ff" if R[i, j] > 0.55 else "#070b1a")
ax.set_title("Correlation Matrix of Housing Features"); ax.grid(False)
plt.colorbar(im, fraction=0.046, pad=0.04)
plt.tight_layout(); plt.show()
```

```python
# 1. Eigendecomposition & Sorting
eigvals, eigvecs = np.linalg.eigh(R)
order = np.argsort(eigvals)[::-1]  # descending order
eigvals, eigvecs = eigvals[order], eigvecs[:, order]

# 2. Variance Metrics & Summary Table
evr = eigvals / eigvals.sum()
summary = pd.DataFrame(
    {
        "Eigenvalue": eigvals,
        "Explained %": evr * 100,
        "Cumulative %": np.cumsum(evr) * 100,
    },
    index=[f"PC{i+1}" for i in range(len(eigvals))],
)
display(summary)

print(
    "Sum of eigenvalues =",
    eigvals.sum().round(4),
    " = trace of R =",
    np.trace(R),
)

# 3. Dynamic Scree Plot Setup
n_comps = len(eigvals)
comp_range = np.arange(1, n_comps + 1)

# Color fallbacks
bar_color = globals().get("GOLD", "#a78bfa")
line_color = globals().get("EMERALD", "#38bdf8")
ref_color = globals().get("ROSE", "#f43f5e")
cum_color = globals().get("TEAL", "#2dd4bf")

fig, ax = plt.subplots(1, 2, figsize=(13, 4.5))

# Plot 1: Scree Plot (Eigenvalues)
ax[0].bar(comp_range, eigvals, color=bar_color, edgecolor="#070b1a", alpha=0.85)
ax[0].plot(comp_range, eigvals, color=line_color, marker="o", lw=2)
ax[0].axhline(1, color=ref_color, ls="--", lw=1.2, label="Kaiser Criterion (λ=1)")
ax[0].set_title("Scree Plot (Eigenvalues)", fontweight="bold")
ax[0].set_xlabel("Principal Component")
ax[0].set_ylabel("Eigenvalue")
ax[0].set_xticks(comp_range)
ax[0].legend()
ax[0].grid(axis="y", linestyle=":", alpha=0.3)

# Plot 2: Cumulative Explained Variance (%)
ax[1].plot(
    comp_range, np.cumsum(evr) * 100, color=cum_color, marker="o", linewidth=2.4
)
ax[1].axhline(80, color=ref_color, ls=":", lw=1, label="80% Threshold")
ax[1].set_title("Cumulative Explained Variance (%)", fontweight="bold")
ax[1].set_xlabel("Number of Components")
ax[1].set_ylabel("Variance Explained (%)")
ax[1].set_xticks(comp_range)
ax[1].set_ylim(0, 105)
ax[1].legend()
ax[1].grid(axis="both", linestyle=":", alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
U, S, Vt = np.linalg.svd(Zs, full_matrices=False)

print("Standardized data  Zs :", Zs.shape)
print("U   (house patterns)  :", U.shape)
print("S   (singular values) :", S.shape, "->", S.round(2))
print("Vt  (feature patterns):", Vt.shape)

# 1) Perfect reconstruction
print("\nU @ diag(S) @ Vt == Zs  :", np.allclose(U @ np.diag(S) @ Vt, Zs))

# 2) Link with eigenvalues of the correlation matrix
print("S^2/n == eigenvalues of R :", np.allclose(S**2 / len(Zs), eigvals))
```

```python
# Low-rank SVD reconstruction & metrics
ranks = range(1, len(S) + 1)
norm_Z = np.linalg.norm(Zs)
total_energy = (S**2).sum()

errors = [
    np.linalg.norm(Zs - (U[:, :k] * S[:k]) @ Vt[:k]) / norm_Z for k in ranks
]
energy = [(S[:k] ** 2).sum() / total_energy * 100 for k in ranks]

# Summary Table
res = pd.DataFrame(
    {"Rank k": ranks, "Error %": np.array(errors) * 100, "Variance Kept %": energy}
).set_index("Rank k")
display(res.round(2))

# Plot
plt.figure(figsize=(7.5, 4))
plt.plot(
    ranks,
    res["Error %"],
    marker="o",
    color="#f43f5e",
    lw=2,
    label="Error %",
)
plt.plot(
    ranks,
    energy,
    marker="s",
    color="#2dd4bf",
    lw=2,
    label="Variance Kept %",
)
plt.title("SVD Low-Rank Approximation", fontweight="bold")
plt.xlabel("Rank k")
plt.xticks(list(ranks))
plt.legend()
plt.tight_layout()
plt.show()
```

```python
model_feats = ["sqft_living", "bedrooms", "bathrooms", "floors", "waterfront",
               "view", "condition", "grade", "yr_built", "sqft_living15"]

Xm = df[model_feats].to_numpy(float)
ym = df["price"].to_numpy(float)

rng = np.random.default_rng(42)
idx = rng.permutation(len(Xm))
cut = int(0.8 * len(Xm))
tr, te = idx[:cut], idx[cut:]

add_bias = lambda M: np.column_stack([M, np.ones(len(M))])
Xtr, Xte = add_bias(Xm[tr]), add_bias(Xm[te])
ytr, yte = ym[tr], ym[te]
print(f"Train: {Xtr.shape}   Test: {Xte.shape}")
```

```python
beta = np.linalg.solve(Xtr.T @ Xtr, Xtr.T @ ytr)

# Cross-check with SVD-based least squares and the pseudo-inverse
beta_ls  = np.linalg.lstsq(Xtr, ytr, rcond=None)[0]
beta_pin = np.linalg.pinv(Xtr) @ ytr
print("solve == lstsq == pinv :", np.allclose(beta, beta_ls) and np.allclose(beta, beta_pin))

coef = pd.Series(beta, index=model_feats + ["intercept"])
coef_table = coef.to_frame("Coefficient ($ per unit)")
display(coef_table)
```

```python
def evaluate(y_true, y_pred):
    ss_res = ((y_true - y_pred) ** 2).sum()
    ss_tot = ((y_true - y_true.mean()) ** 2).sum()
    return {"R2": 1 - ss_res / ss_tot,
            "RMSE": np.sqrt(np.mean((y_true - y_pred) ** 2)),
            "MAE": np.mean(np.abs(y_true - y_pred))}

# Three models on the SAME test set
base_pred = np.full_like(yte, ytr.mean())                         # 1) always predict the average
simple_cols = [model_feats.index("sqft_living"), model_feats.index("bedrooms")]
Xs_tr = np.column_stack([Xm[tr][:, simple_cols], np.ones(len(tr))])
Xs_te = np.column_stack([Xm[te][:, simple_cols], np.ones(len(te))])
simple_pred = Xs_te @ np.linalg.solve(Xs_tr.T @ Xs_tr, Xs_tr.T @ ytr)   # 2) price = a*sqft + b*bed + c
full_pred = Xte @ beta                                                    # 3) full matrix model

scores_tbl = pd.DataFrame({"Average only": evaluate(yte, base_pred),
                           "sqft + bedrooms": evaluate(yte, simple_pred),
                           "Full matrix model": evaluate(yte, full_pred)}).T
display(scores_tbl)

m = evaluate(yte, full_pred)
r2_test, rmse_test, mae_test = m["R2"], m["RMSE"], m["MAE"]

# Optional verification against scikit-learn
try:
    from sklearn.linear_model import LinearRegression
    sk = LinearRegression().fit(Xm[tr], ytr)
    print("\nscikit-learn gives the same R2:", np.isclose(sk.score(Xm[te], yte), r2_test))
except ImportError:
    print("\n(scikit-learn not installed - skipped optional check)")
```
